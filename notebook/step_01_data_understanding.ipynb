# Customer Churn Prediction — Step 1: Data Understanding

**Dataset:** Telco Customer Churn (Kaggle / IBM Watson Analytics)  
**Goal:** Load the raw dataset and perform an initial inspection — no cleaning, no transformation, no modelling.

---

## 1. Import Libraries

We only need **pandas** for this step.  
`pandas` is the standard Python library for loading, inspecting, and manipulating tabular data.  
We also import `os` so we can build a file path that works regardless of where the notebook is run from.

```python
import os
import pandas as pd
```

---
## 2. Define the Dataset Path

`os.path.join` builds an OS-agnostic file path.  
`os.path.dirname(__file__)` would give us the notebook's own directory, but inside Jupyter the equivalent is to navigate relative to the notebook location.  
Here we go **one level up** from `notebook/` to reach the project root, then down into `data/`.  
This makes the path work whether you run the notebook from any directory.

```python
# Build a path relative to the notebook's location so it works on any machine
NOTEBOOK_DIR = os.path.abspath('')          # current working directory when running this cell
PROJECT_ROOT = os.path.dirname(NOTEBOOK_DIR) if os.path.basename(NOTEBOOK_DIR) == 'notebook' else NOTEBOOK_DIR
DATA_PATH = os.path.join(PROJECT_ROOT, 'data', 'WA_Fn-UseC_-Telco-Customer-Churn.csv')

print('Dataset path:', DATA_PATH)
print('File exists :', os.path.exists(DATA_PATH))
```

---
## 3. Load the Dataset

`pd.read_csv()` reads a comma-separated values file and returns a **DataFrame** — pandas' core 2-D table structure.  
No arguments beyond the file path are passed here because we want to load the data **exactly as it is** in the raw file without applying any parsing tricks or type coercions.

```python
df = pd.read_csv(DATA_PATH)

print('Dataset loaded successfully.')
```

---
## 4. First 5 Rows — `df.head()`

`df.head(5)` returns the **first 5 rows** of the DataFrame as a formatted table.  
This gives an immediate visual check that:
- The file was read correctly.
- Column headers were parsed from the first row.
- Values look as expected (no obvious encoding issues, correct delimiters, etc.).

```python
df.head(5)
```

---
## 5. Dataset Shape — `df.shape`

`df.shape` returns a tuple **(rows, columns)**.  
- **Rows** = the number of customer records in the dataset.  
- **Columns** = the number of features (variables) available for each customer.  

Knowing the dimensions upfront helps us understand the scale of the data before doing anything else.

```python
rows, cols = df.shape

print(f'Number of rows    (customer records) : {rows}')
print(f'Number of columns (features)         : {cols}')
```

---
## 6. Column Names — `df.columns`

`df.columns` returns an **Index** of all column names.  
Printing them lets us:
- Confirm every expected feature is present.
- Spot any typos or unusual naming conventions.
- Plan ahead for future cleaning and feature engineering steps.

We enumerate them with their index numbers so they are easier to reference.

```python
print(f'Total columns: {len(df.columns)}\n')
print('Column index  |  Column name')
print('-' * 40)
for i, col in enumerate(df.columns):
    print(f'  {i:>2}          |  {col}')
```

---
## 7. Quick Summary — `df.info()`

`df.info()` prints a concise summary that shows, for every column:
- The **data type** pandas inferred (`int64`, `float64`, `object`, etc.).
- The **non-null count** — any column with fewer non-null entries than total rows has missing values.

This is purely observational here. We are **not fixing anything yet** — we are only noting what we see for the data quality analysis that comes in Step 2.

```python
df.info()
```

---
## 8. Basic Descriptive Statistics — `df.describe()`

`df.describe()` computes summary statistics (count, mean, std, min, quartiles, max) for every **numeric** column.  
Passing `include='all'` also shows basic stats for **categorical / object** columns (unique count, top value, frequency).  

Again — purely observational. Nothing is changed.

```python
df.describe(include='all')
```
